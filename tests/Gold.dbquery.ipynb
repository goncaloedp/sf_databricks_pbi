```sql
SELECT *
FROM gold.eventos
LIMIT 25;

SELECT COUNT(*) AS total_rows
FROM gold.ativos;

SELECT salesforce_id,
       COUNT(*) n
FROM gold.ativos
GROUP BY salesforce_id
HAVING COUNT(*) > 1;

SELECT Country__c,
       COUNT(*)
FROM gold.ativos
GROUP BY Country__c;

SELECT Lote,
       COUNT(*)
FROM gold.ativos
GROUP BY Lote
ORDER BY Lote;

SELECT *
FROM gold.eventos
WHERE Days_Task_Owner < 0;

SELECT COUNT(*)
FROM gold.eventos
WHERE salesforce_id IS NULL;

SELECT case_id,
COUNT(*)
FROM gold.pedidos_operacao
GROUP BY case_id
HAVING COUNT(*)>1;

SELECT Status,
COUNT(*)
FROM gold.pedidos_operacao
GROUP BY Status;

SELECT COUNT(*)
FROM gold.empresa
WHERE nome_empresa IS NULL;

SELECT COUNT(*)
FROM gold.eventos e
LEFT JOIN gold.ativos a
ON e.salesforce_id=a.salesforce_id
WHERE a.salesforce_id IS NULL;

SELECT COUNT(*)
FROM gold.pedidos_operacao p
LEFT JOIN gold.eventos e
ON p.ParentId=e.case_id
WHERE e.case_id IS NULL;
```
